### OEF 8: Konijen en vossen

Veronderstel een gesloten gebied waar konijnen en vossen samen leven. De volledige konijnenpopulatie stellen we voor door $k(t)$, de volledige vossenpopulatie stellen we voor door $v(t)$. Om de evolutie van de konijnen- en vossenpopulatie te begrijpen (en te voorspellen), moeten we met volgende zaken rekening houden:

- **De geboorte van nieuwe konijnen.** Dit gebeurt zoals het exponentiële groeimodel uit het voorbeeld. Konijnen vinden altijd voldoende gras en kunnen zich zonder jagers onbegrensd voortplanten. De sterfte van konijnen is verwaarloosbaar ten opzichte van het hoge aantal geboortes. Als de groeifactor $a$ is, dan wordt het groeimodel:
$$ \frac{dk(t)}{dt} = a \ k(t)$$

- **Predatie door vossen.** Door de aanwezigheid van vossen wordt de groei van de konijnenpopulatie echter tegengehouden. Hoeveel die groei wordt tegengehouden, dat hangt af van het aantal vossen (hoe meer vossen, hoe moeilijker de konijnenpopulatie kan groeien), maar ook van het aantal konijnen (hoe meer konijnen, des te gemakkelijker die ten prooi kunnen vallen aan de vossen). In ons model vermenigvuldigen we beide invloeden met elkaar en wegen die opnieuw met een parameter $b$. Voegen we dit toe aan het bovenstaande model, dan bekomen we:
$$ \boxed{\frac{dk(t)}{dt} = a \ k(t) - b \ k(t) \ v(t)}$$
Let hierbij zeker op het minteken dat verduidelijkt dat de tweede term geen groei (+) maar wel een afname (-) teweegbrengt in de konijnenpopulatie.

- **Jacht op konijnen.** De groei van de vossenpopulatie is uiteraard afhankelijk van het aantal vossen, maar ook van het aantal konijnen: als er te weinig konijnen zijn, heeft de vossenpopulatie te weinig voedsel en zullen ze ook geen nakomelingen krijgen. We introduceren hiervoor een productterm met een groeifactor $d$ en toegevoegd in de differentiaalvergelijking:
$$ \frac{dv(t)}{dt} = d \ k(t) \ v(t) $$

- **Concurrentie tussen vossen.** In tegenstelling tot bij de konijnen, is sterfte binnen de vossenpopulatie wel van belang: roofdieren vertonen namelijk territoriaal gedrag en concurreren met elkaar voor voedsel. Dit komt overeen met een negatieve term, waardoor de evolutie van de vossenpopulatie gemodelleerd wordt als volgt:
$$ \boxed{\frac{dv (t)}{dt} = -c \ k(t) + d \ k(t) \ v(t)}$$

Nu trekken we de massa 0.20 meter uit zijn evenwichtspositie en laten die van daaruit zonder beginsnelheid los. Na het loslaten ondervindt de massa enkel de veerkracht $F_v = k x$. Via de tweede wet van Newton (zie fysica 2e semester) vinden we volgende differentiaalvergelijking die de beweging van de massa beschrijft: 

$$ m \frac{d^2x(t)}{dt^2} = -k \cdot x(t) $$

Hierin is:
- $m$ = 0.1 kg de massa,
- $k$ = 5.0 N/m de veerconstante,
- $x(t)$ de positie ten opzichte van de evenwichtsstand, evoluerend in de tijd

Omdat bovenstaande differentiaalvergelijking een afgeleide bevat van de tweede orde, is de discretisatie zoals hierboven beschreven niet mogelijk. Toch is er een slimme manier om dit probleem alsnog op te lossen. Daarvoor introduceren we een tweede differentiaalvergelijking en lossen die twee samen op:
$$
\begin{cases}
\frac{dx(t)}{dt} &= v(t) \\
\frac{dv(t)}{dt} &= -\frac{k}{m} \cdot x(t) \\
\end{cases}
$$

### OEF 10: SIR-model als epidemiologisch model

In OEF 6 modelleerde je de verspreiding van een ziekte door de evolutie van het aantal besmette personen te bepalen. Hierbij hield je geen rekening met genezing, waardoor uiteindelijk de volledige populatie besmet is geweest.

Een meer realistisch model is het *SIR-model*, waarbij de letters staan voor Susceptible (vatbaar), Infectious (besmettelijk) en Recovered (genezen). Hierin worden de volgende aannames gemaakt:
* De vatbare personen $S$ worden besmet met een snelheid evenredig met hun aantal $S$ en het aantal besmette personen $I$ . Het aantal besmette personen neemt met dezelfde snelheid toe.
* De besmette personen $I$ genezen met een snelheid evenredig met hun aantal $I$. Het aantal genezen personen neemt met dezelfde snelheid toe.
Op die manier bekom je het volgende stelsel differentiaalvergelijkingen.

$$\begin{align*}
    \frac{dS}{dt} &= -b S I \\
    \frac{dI}{dt} &= b S I - c I \\
    \frac{dR}{dt} &= c I \\
\end{align*}$$

Merk op dat het totale bevolkingsaantal niet verandert in de tijd. Er geldt immers dat
$$
\frac{d}{dt}(S + I + R) = -b S I + b S I - c I + c I = 0
$$
